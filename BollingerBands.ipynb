Here is the formula for the 3 lines.

Middle Band = 20 day simple moving average (SMA)

Upper Band = 20 day SMA + (20-day standard deviation of price x 2) 

Lower Band = 20 day SMA – (20-day standard deviation of price x 2)

How to calculate Bollinger Bands?
Traders use the formulas to calculate the three deviation:

How to calculate Bollinger Bands?
Standard deviation (SD) is the square root of the summation of the numbers in the population minus the mean divided by sample size. The next step after calculating the Standard deviation is calculating the upper and lower Bollinger bands. The Bollinger upper and lower bands are calculated by multiplying SD by two and both adding and subtracting the number from the value to plot upper and lower values.

The Bollinger Upper band is calculated by adding the 20-day simple moving average with the 20-day standard deviation multiplied by 2. Here is the formula for calculating the Bollinger Upper band:

BOLU=MA(TP,n)+m∗σ[TP,n]

The Bollinger Lower Band is calculated by subtracting the 20-day simple moving average from the 20-day standard deviation multiplied by 2. Here is the formula for calculating the Bollinger lower band:

BOLD=MA(TP,n)−m∗σ[TP,n]

Where,

BLOU = Bollinger Upper Band

BOLD = Bollinger Lower Band

MA = Moving Average

TP= Typical Price (High+Low+Close) / 3

N = Number of days in Moving Average (typically 20)

M = Number of SD (typically 2)

σ[TP,n] = SD over last n periods of TP

```python
def BollingerBands(data, ma_period, n_std=2):
    
    sma= data.Close.rolling(ma_period).mean()
    
    hlc_avg= (data.High + data.Low + data.Close)/3
    
    # We calculated MA(TP,n):
    mean= hlc_avg.rolling(ma_period).mean()
    
    # calculated the standard deviation σ[TP,n]:
    std= hlc_avg.rolling(ma_period).std()
    
    # upper band
    upper_band= mean + std * n_std
    
    # lower band
    lower_band= mean - std * n_std
    
    return sma, upper_band, lower_band
```

```python
df["SMA"], df["BBUp"], df["BBDown"]= BollingerBands(df, 20, 2)
df
```

```python
def backtest_bollingerBands(df, rsi_threshold_low= 30, rsi_threshold_high= 70, bb_width_threshold= 0.0015):
    df['Signal']= 0
    for i in range(1, len(df)):
        #Previous Candle Conditions for Long Signal:
        prev_candle_closes_below_bb= df['Close'].iloc[i-1] < df['BBDown'].iloc[i-1]
        prev_rsi_below_thr= df['RSI'].iloc[i-1] < rsi_threshold_low
        #Current Candle
        closes_above_prev_high= df['Close'].iloc[i] > df['High'].iloc[i-1]
        bb_width_greater_threshold= df['BbWidth'].iloc[i] > bb_width_threshold
        #SMA trend
        #price_moves_above_the_SMA= df['Close'].iloc[i] > df['SMA'].iloc[i]

        #Combine Conditions
        if(prev_candle_closes_below_bb and
          prev_rsi_below_thr and
          closes_above_prev_high and
          bb_width_greater_threshold 
          # and price_moves_above_the_SMA
          ):
            df.at[df.index[i],'Signal']= 1 #Buy Signal
        
        #Previous Candle Conditions for Short Signal:
        prev_candle_closes_above_bb= df['Close'].iloc[i-1] > df['BBUp'].iloc[i-1]
        prev_rsi_above_thr= df['RSI'].iloc[i-1] > rsi_threshold_high
        #Current Candle
        closes_below_prev_low= df['Close'].iloc[i] < df['Low'].iloc[i-1]
        bb_width_greater_threshold= df['BbWidth'].iloc[i] > bb_width_threshold
        #SMA trend
        #price_moves_below_the_SMA= df['Close'].iloc[i] < df['SMA'].iloc[i]

        #Combine Conditions
        if(prev_candle_closes_above_bb and
          prev_rsi_above_thr and
          closes_below_prev_low and
          bb_width_greater_threshold 
          # and price_moves_below_the_SMA
          ):
            df.at[df.index[i],'Signal']= -1 #Sell Signal
    return df
```

```python
backtest_bollingerBands(signal_df, 30, 70, 0.03)
# it gets very selective with bb_width_threshold as 0.03
```
